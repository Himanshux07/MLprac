```python
import time
from collections import deque


# =========================================================
# BFS for Graph
# =========================================================
def bfs_graph(graph, start, goal):
    queue = deque()
    queue.append((start, [start]))

    visited = set()
    nodes_explored = 0

    while queue:
        current, path = queue.popleft()

        if current in visited:
            continue

        visited.add(current)
        nodes_explored += 1

        if current == goal:
            return path, nodes_explored

        for neighbor in graph[current]:
            if neighbor not in visited:
                queue.append((neighbor, path + [neighbor]))

    return None, nodes_explored


# =========================================================
# DFS for Graph
# =========================================================
def dfs_graph(graph, start, goal):
    stack = []
    stack.append((start, [start]))

    visited = set()
    nodes_explored = 0

    while stack:
        current, path = stack.pop()

        if current in visited:
            continue

        visited.add(current)
        nodes_explored += 1

        if current == goal:
            return path, nodes_explored

        for neighbor in reversed(graph[current]):
            if neighbor not in visited:
                stack.append((neighbor, path + [neighbor]))

    return None, nodes_explored


# =========================================================
# Path Finding using Nodes and Edges
# =========================================================
def path_finding():

    print("\n========== PATH FINDING ==========")

    n = int(input("Enter number of nodes: "))

    if n <= 0:
        print("Number of nodes must be greater than 0.")
        return

    max_edges = n * (n - 1) // 2

    e = int(input("Enter number of edges: "))

    if e < 0 or e > max_edges:
        print("Number of edges must be between 0 and", max_edges)
        return

    graph = {}

    for i in range(n):
        graph[i] = []

    print("\nEnter edges.")
    print("Node numbers must be between 0 and", n - 1)

    edges = set()

    for i in range(e):

        while True:

            try:
                u, v = map(
                    int,
                    input("Edge " + str(i + 1) + ": ").split()
                )

                if u < 0 or u >= n or v < 0 or v >= n:
                    print("Invalid node. Enter nodes between 0 and", n - 1)
                    continue

                if u == v:
                    print("A node cannot be connected to itself.")
                    continue

                edge = tuple(sorted((u, v)))

                if edge in edges:
                    print("This edge already exists. Enter another edge.")
                    continue

                edges.add(edge)

                graph[u].append(v)
                graph[v].append(u)

                break

            except ValueError:
                print("Please enter two valid node numbers.")

    # Starting node
    start = int(input("\nEnter starting node: "))

    while start < 0 or start >= n:
        print("Starting node must be between 0 and", n - 1)
        start = int(input("Enter starting node: "))

    # Destination node
    goal = int(input("Enter destination node: "))

    while goal < 0 or goal >= n:
        print("Destination node must be between 0 and", n - 1)
        goal = int(input("Enter destination node: "))

    # =====================================================
    # BFS
    # =====================================================

    start_time = time.perf_counter()

    bfs_path, bfs_nodes = bfs_graph(graph, start, goal)

    bfs_time = time.perf_counter() - start_time

    print("\n------------- BFS -------------")

    if bfs_path:
        print("Path:", " -> ".join(map(str, bfs_path)))
        print("Path Length:", len(bfs_path) - 1)
    else:
        print("No path found")
        print("Path Length: 0")

    print("Nodes Explored:", bfs_nodes)
    print("Execution Time:", bfs_time, "seconds")

    # =====================================================
    # DFS
    # =====================================================

    start_time = time.perf_counter()

    dfs_path, dfs_nodes = dfs_graph(graph, start, goal)

    dfs_time = time.perf_counter() - start_time

    print("\n------------- DFS -------------")

    if dfs_path:
        print("Path:", " -> ".join(map(str, dfs_path)))
        print("Path Length:", len(dfs_path) - 1)
    else:
        print("No path found")
        print("Path Length: 0")

    print("Nodes Explored:", dfs_nodes)
    print("Execution Time:", dfs_time, "seconds")


# =========================================================
# Get Neighbours of a Maze Cell
# =========================================================
def get_neighbors(row, col, rows, cols):

    directions = [
        (-1, 0),   # Up
        (1, 0),    # Down
        (0, -1),   # Left
        (0, 1)     # Right
    ]

    neighbors = []

    for dr, dc in directions:

        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < rows and 0 <= new_col < cols:
            neighbors.append((new_row, new_col))

    return neighbors


# =========================================================
# BFS for Maze
# =========================================================
def bfs_maze(maze, start, goal):

    rows = len(maze)
    cols = len(maze[0])

    queue = deque()
    queue.append((start, [start]))

    visited = set()
    nodes_explored = 0

    while queue:

        current, path = queue.popleft()

        if current in visited:
            continue

        visited.add(current)
        nodes_explored += 1

        if current == goal:
            return path, nodes_explored

        row, col = current

        for nr, nc in get_neighbors(row, col, rows, cols):

            if maze[nr][nc] != '#' and (nr, nc) not in visited:
                queue.append(
                    ((nr, nc), path + [(nr, nc)])
                )

    return None, nodes_explored


# =========================================================
# DFS for Maze
# =========================================================
def dfs_maze(maze, start, goal):

    rows = len(maze)
    cols = len(maze[0])

    stack = []
    stack.append((start, [start]))

    visited = set()
    nodes_explored = 0

    while stack:

        current, path = stack.pop()

        if current in visited:
            continue

        visited.add(current)
        nodes_explored += 1

        if current == goal:
            return path, nodes_explored

        row, col = current

        for nr, nc in get_neighbors(row, col, rows, cols):

            if maze[nr][nc] != '#' and (nr, nc) not in visited:
                stack.append(
                    ((nr, nc), path + [(nr, nc)])
                )

    return None, nodes_explored


# =========================================================
# Print Maze Path
# =========================================================
def print_maze(maze, path):

    new_maze = [row[:] for row in maze]

    if path:

        for row, col in path:

            if new_maze[row][col] != 'S' and new_maze[row][col] != 'E':
                new_maze[row][col] = '*'

    print()

    for row in new_maze:
        print(" ".join(row))


# =========================================================
# Maze Navigation
# =========================================================
def maze_navigation():

    print("\n========== MAZE NAVIGATION ==========")

    rows = int(input("Enter number of rows: "))
    cols = int(input("Enter number of columns: "))

    if rows <= 0 or cols <= 0:
        print("Rows and columns must be greater than 0.")
        return

    maze = []

    print("\nEnter maze:")
    print("S = Start")
    print("E = End")
    print(". = Open path")
    print("# = Wall")

    # -----------------------------------------------------
    # Input maze row by row
    # -----------------------------------------------------

    for i in range(rows):

        while True:

            row = input("Row " + str(i + 1) + ": ").split()

            if len(row) != cols:
                print("Please enter exactly", cols, "values.")
            else:
                maze.append(row)
                break

    # -----------------------------------------------------
    # Find Start and End
    # -----------------------------------------------------

    start = None
    goal = None

    for i in range(rows):

        for j in range(cols):

            if maze[i][j] == 'S':
                start = (i, j)

            elif maze[i][j] == 'E':
                goal = (i, j)

    # Check Start
    if start is None:
        print("\nStart 'S' is missing.")
        return

    # Check End
    if goal is None:
        print("\nEnd 'E' is missing.")
        return

    # =====================================================
    # BFS
    # =====================================================

    start_time = time.perf_counter()

    bfs_path, bfs_nodes = bfs_maze(
        maze,
        start,
        goal
    )

    bfs_time = time.perf_counter() - start_time

    print("\n------------- BFS -------------")

    if bfs_path:

        print("Path Length:", len(bfs_path) - 1)
        print("Nodes Explored:", bfs_nodes)
        print("Execution Time:", bfs_time, "seconds")

        print("\nBFS Path:")
        print_maze(maze, bfs_path)

    else:

        print("No path found")
        print("Nodes Explored:", bfs_nodes)

    # =====================================================
    # DFS
    # =====================================================

    start_time = time.perf_counter()

    dfs_path, dfs_nodes = dfs_maze(
        maze,
        start,
        goal
    )

    dfs_time = time.perf_counter() - start_time

    print("\n------------- DFS -------------")

    if dfs_path:

        print("Path Length:", len(dfs_path) - 1)
        print("Nodes Explored:", dfs_nodes)
        print("Execution Time:", dfs_time, "seconds")

        print("\nDFS Path:")
        print_maze(maze, dfs_path)

    else:

        print("No path found")
        print("Nodes Explored:", dfs_nodes)


# =========================================================
# Main Menu
# =========================================================

while True:

    print("\n======================================")
    print("       BFS AND DFS ROUTE FINDING")
    print("======================================")
    print("1. Path Finding (Nodes and Edges)")
    print("2. Maze Navigation")
    print("3. Exit")
    print("======================================")

    choice = input("Enter your choice: ")

    if choice == '1':

        path_finding()

    elif choice == '2':

        maze_navigation()

    elif choice == '3':

        print("\nProgram ended.")
        break

    else:

        print("\nInvalid choice. Please try again.")
```
